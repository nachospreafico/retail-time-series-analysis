```python
valid_sale = (
    (df["Quantity"] > 0) &
    (df["Price"] > 0)
)
```

```python
sales = (
    df
    .drop_duplicates()
    .loc[
        (df["Quantity"] > 0) &
        (df["Price"] > 0)
    ]
    .copy()
)

sales["Revenue"] = sales["Quantity"] * sales["Price"]
```

```python
print("Raw rows:", len(df))
print("Sales rows:", len(sales))
print("Rows removed:", len(df) - len(sales))
```

```python
sales[["Quantity", "Price", "Revenue"]].describe()
```

```python
sales.groupby("Country")["Revenue"].sum().sort_values(ascending=False).head(15)
```

```python
uk_share = (
    sales.loc[sales["Country"] == "United Kingdom", "Revenue"].sum()
    / sales["Revenue"].sum()
)

uk_share
```

```python
uk_sales = sales[
    sales["Country"] == "United Kingdom"
].copy()
```

```python
uk_sales["InvoiceDate"].min(), uk_sales["InvoiceDate"].max()
```

```python
uk_sales["Date"] = uk_sales["InvoiceDate"].dt.normalize()
```

```python
daily_revenue = (
    uk_sales
    .groupby("Date")["Revenue"]
    .sum()
    .sort_index()
)

daily_revenue.head()
```

```python
daily_revenue.tail()
```

```python
daily_revenue.index.nunique()
```

```python
full_date_range = pd.date_range(
    daily_revenue.index.min(),
    daily_revenue.index.max(),
    freq="D"
)

len(full_date_range)
```

```python
daily_revenue_df = daily_revenue.to_frame(name="Revenue")

daily_revenue_df["day_of_week"] = daily_revenue_df.index.day_name()

daily_revenue_df.groupby("day_of_week")["Revenue"].agg(
    ["count", "mean", "median"]
)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(14, 5))

daily_revenue.plot(ax=ax)

ax.set_title("UK Daily Revenue")
ax.set_xlabel("Date")
ax.set_ylabel("Revenue (£)")

plt.show()
```
